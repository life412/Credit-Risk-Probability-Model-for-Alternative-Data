```python
import matplotlib.pyplot as plt
import seaborn as sns

num_cols = ['Amount', 'Value', 'CountryCode', 'PricingStrategy']

plt.figure(figsize=(12,8))
for i, col in enumerate(num_cols):
    plt.subplot(2, 2, i+1)
    sns.histplot(df[col], kde=True)
    plt.title(f"Distribution of {col}")

plt.tight_layout()
plt.show()
```

```python
cat_cols = ['CurrencyCode', 'ProductCategory', 'ChannelId', 'FraudResult']

for col in cat_cols:
    plt.figure(figsize=(8,4))
    sns.countplot(data=df, x=col)
    plt.title(f"Count Plot of {col}")
    plt.xticks(rotation=45)
    plt.show()
```

## EDA Summary

### 1. Dataset Overview
- The dataset contains multiple transaction records with fields like TransactionId, CustomerId, Amount, Value, ProductCategory, ChannelId, and FraudResult.
- Most columns are either categorical or numeric.
- Missing values are minimal, so the dataset is generally clean.

### 2. Numerical Feature Analysis
- Amount and Value are highly skewed; a few transactions are extremely high.
- Negative Amount values indicate credits, positive values are debits.
- CountryCode is mostly 256; PricingStrategy has few discrete values.

### 3. Categorical Feature Analysis
- CurrencyCode is almost always UGX → low predictive power.
- ProductCategory (airtime, financial_services, utility_bill) and ChannelId show meaningful variation — potential predictive features.
- FraudResult is imbalanced (mostly 0) → may need careful handling if used as a feature.

### 4. Correlation Analysis
- Amount and Value are strongly correlated.
- Other numeric features show weak correlation with FraudResult.

### 5. Key Insights
1. Most transactions are small, but a few are extremely large — skewed distribution.
2. Negative Amount values represent credits; need careful treatment during modeling.
3. ProductCategory and ChannelId show variation across users → good predictors.
4. FraudResult is highly imbalanced → cannot be used directly as a label.
5. Behavioral features like RFM will be necessary to create a proxy target variable.
